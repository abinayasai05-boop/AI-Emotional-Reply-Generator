```python
# ============================================================
# AI EMOTIONAL REPLY GENERATOR
# Google Colab - Complete Version
# ============================================================

import random
import ipywidgets as widgets
from IPython.display import display, Markdown
from transformers import pipeline

# ------------------------------------------------------------
# Load AI Emotion Model
# ------------------------------------------------------------

print("⏳ Loading AI emotion detection model...")

emotion_classifier = pipeline(
    "text-classification",
    model="j-hartmann/emotion-english-distilroberta-base",
    top_k=None
)

print("✅ AI model loaded successfully!")


# ------------------------------------------------------------
# Emotional Reply Database
# ------------------------------------------------------------

emotion_replies = {

    "anger": [
        "I can understand that you're feeling angry. Take a moment to breathe before reacting.",
        "It sounds like something really frustrated you. Your feelings are understandable.",
        "I understand why you might feel upset. Taking some time to calm down could help."
    ],

    "sadness": [
        "I'm sorry you're feeling this way. Be gentle with yourself during difficult moments.",
        "It sounds like you're going through a difficult time. Take things one step at a time.",
        "I'm here to listen. Sometimes sharing what you're feeling can make things a little easier."
    ],

    "joy": [
        "That's wonderful to hear! I'm really happy for you.",
        "You sound really happy! It's great to celebrate moments like this.",
        "That's fantastic! I hope you continue enjoying this positive moment."
    ],

    "fear": [
        "It sounds like you're feeling worried. Take a slow breath and focus on what you can control.",
        "I understand that this situation may feel frightening. Take things one step at a time.",
        "It seems like you're feeling anxious. Give yourself a moment to process everything."
    ],

    "surprise": [
        "Wow, that sounds unexpected! I can understand why you're surprised.",
        "That definitely sounds like something you weren't expecting!",
        "It sounds like something caught you completely off guard."
    ],

    "disgust": [
        "It sounds like that situation really bothered you. Your reaction is understandable.",
        "I can understand why you might feel uncomfortable about that.",
        "That doesn't sound like a pleasant experience."
    ],

    "neutral": [
        "I understand. Thanks for sharing that with me.",
        "I hear you. Tell me more about what you're thinking.",
        "Thanks for explaining. I'm here to listen."
    ]
}


# ------------------------------------------------------------
# AI Reply Generator
# ------------------------------------------------------------

def generate_emotional_reply(text):

    if not text or not text.strip():
        return None, 0, "Please enter a message."

    # Get emotion predictions
    results = emotion_classifier(text)[0]

    # Sort by confidence
    results = sorted(
        results,
        key=lambda x: x["score"],
        reverse=True
    )

    # Highest confidence emotion
    emotion = results[0]["label"].lower()

    confidence = results[0]["score"]

    # Generate response
    reply = random.choice(
        emotion_replies.get(
            emotion,
            emotion_replies["neutral"]
        )
    )

    return emotion, confidence, reply


# ------------------------------------------------------------
# Interactive UI
# ------------------------------------------------------------

text_box = widgets.Textarea(
    value="",
    placeholder="Type something you're feeling...",
    description="Message:",
    layout=widgets.Layout(
        width="100%",
        height="120px"
    )
)

generate_button = widgets.Button(
    description="🤖 Generate Emotional Reply",
    button_style="success",
    layout=widgets.Layout(
        width="250px",
        height="40px"
    )
)

clear_button = widgets.Button(
    description="🗑️ Clear",
    button_style="warning"
)

output = widgets.Output()


# ------------------------------------------------------------
# Generate Button Function
# ------------------------------------------------------------

def generate_reply_clicked(button):

    with output:

        output.clear_output()

        user_text = text_box.value.strip()

        if not user_text:
            print("⚠️ Please enter a message first.")
            return

        emotion, confidence, reply = generate_emotional_reply(
            user_text
        )

        print("=" * 65)
        print("          🤖 AI EMOTIONAL REPLY GENERATOR")
        print("=" * 65)

        print("\n👤 USER MESSAGE")
        print(user_text)

        print("\n🎭 DETECTED EMOTION")
        print(emotion.upper())

        print("\n📊 CONFIDENCE")
        print(f"{confidence:.2%}")

        print("\n💬 AI EMOTIONAL REPLY")
        print(reply)

        print("\n" + "=" * 65)


# ------------------------------------------------------------
# Clear Button Function
# ------------------------------------------------------------

def clear_clicked(button):

    text_box.value = ""

    with output:
        output.clear_output()


# Connect buttons
generate_button.on_click(generate_reply_clicked)
clear_button.on_click(clear_clicked)


# ------------------------------------------------------------
# Display Application
# ------------------------------------------------------------

display(
    Markdown("# 🤖 AI Emotional Reply Generator")
)

display(
    Markdown(
        "Enter a message below. The AI will detect the emotion "
        "and generate an appropriate emotional reply."
    )
)

display(text_box)

display(
    widgets.HBox([
        generate_button,
        clear_button
    ])
)

display(output)
```
